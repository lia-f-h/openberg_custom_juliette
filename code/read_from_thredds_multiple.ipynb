```python
import xarray as xr
import numpy as np
```

```python
#xr.open_mfdataset('https://thredds.met.no/thredds/catalog/accibergt5/catalog.xml')
ds = xr.open_mfdataset(['https://thredds.met.no/thredds/dodsC/accibergt42/topaz4_be_mem0%s.ncml'%membnr for membnr in [('0'+str(memb)) 
                        if memb<10 else str(memb) for memb in range(1,11)]],
                         combine="nested",
    concat_dim="realization",#should be named "realization" or "ensemble_member"
    parallel=True,
    engine="netcdf4")
ds
```

```python
# create CF/OpenDrift-compliant realization coordinate
ds = ds.assign_coords(
    realization=xr.DataArray(
        ds.realization,
        dims=("realization",),
        attrs={
            "standard_name": "realization",
            "long_name": "ensemble member",
            "axis": "E"
        }
    )
)
ds
```
