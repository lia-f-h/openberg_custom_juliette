```python
import xarray as xr
import numpy as np
```

```python
#xr.open_mfdataset('https://thredds.met.no/thredds/catalog/accibergt5/catalog.xml')
path = ['https://thredds.met.no/thredds/dodsC/accibergt42/topaz4_be_mem0%s.ncml'%membnr for membnr in [('0'+str(memb)) 
                        if memb<10 else str(memb) for memb in range(1,11)]]
with xr.open_mfdataset(
    path,
    combine="nested",
    concat_dim="realization",
    parallel=True,
    engine="netcdf4",
    chunks={'time': 10}
) as ds:

    ds = ds.assign_coords(
        realization=xr.DataArray(
            ds.realization,
            dims=("realization",),
            attrs={
                "standard_name": "realization",
                "long_name": "ensemble member",
                "axis": "E"
            }
        )
    )

ds
```
