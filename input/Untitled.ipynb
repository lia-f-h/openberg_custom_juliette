Download custom arcmfcwam data without currents

```python
import xarray as xr

def read_cmems_custom_variables(dataset_id_in, variable_list_in):
    import copernicusmarine as cm
    
    ds = cm.open_dataset(
        dataset_id=dataset_id_in,
        variables=variable_list_in,
        minimum_longitude=-80,
        maximum_longitude=-50,
        minimum_latitude=70,
        maximum_latitude=80,
        start_datetime="2025-09-01T00:00:00",
        end_datetime="2025-12-31T00:00:00"
    )
    return ds

ds_custom = read_cmems_custom_variables("dataset-wam-arctic-1hr3km-be",["VHM0","VMDR","VSDX","VSDY"])
```
